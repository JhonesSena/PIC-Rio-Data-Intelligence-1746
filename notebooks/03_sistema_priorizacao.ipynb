# Parte 3: Sistema de Priorização

Desenvolvimento de um score multidimensional para otimização de recursos municipais.

### 1. Fórmula do Score de Prioridade (Q9)
Alinhado com a governança da PCRJ:
$$Score = (P_{atraso} \times 0.4) + (U_{urgencia} \times 0.3) + (C_{clima} \times 0.15) + (V_{equidade} \times 0.15)$$

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
df_test = pd.DataFrame({
    'id_chamado': range(1000),
    'prob_atraso': np.random.uniform(0, 1, 1000),
    'urgencia': np.random.choice([0.2, 0.8], 1000),
    'vulnerabilidade': np.random.uniform(0, 1, 1000),
    'atraso_real': np.random.choice([0, 1], 1000, p=[0.8, 0.2])
})

# Cálculo do Score (Q9)
df_test['score'] = (df_test['prob_atraso']*0.4 + df_test['urgencia']*0.3 + df_test['vulnerabilidade']*0.15 + 0.15)

print("Top 5 Prioridades:")
display(df_test.sort_values('score', ascending=False).head())
```

## 2. Simulação de Impacto (Q10)
Comparação entre seleção aleatória vs Priority Score.

```python
top_200_random = df_test.sample(200)['atraso_real'].sum()
top_200_score = df_test.sort_values('score', ascending=False).head(200)['atraso_real'].sum()

plt.bar(['Aleatório', 'Priority Score'], [top_200_random, top_200_score])
plt.title('Atrasos Capturados (Top 20% do Volume)')
plt.show()

print(f"Ganho de Efetividade: {((top_200_score/top_200_random)-1)*100:.2f}%")
```
